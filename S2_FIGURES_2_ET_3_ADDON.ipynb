# **Figure 2 : UMAP Multi-Panel (4 Colorations)**

📍 **Localisation** : Après `sc.tl.umap(adata_pp)` dans S2

Visualisation UMAP avec 4 panels :
- Panel A : Clusters Leiden
- Panel B : Diagnostic (AD/PD/CTRL)
- Panel C : Donneurs (batch effect)
- Panel D : Ascendance génétique

**Output** : `Fig2_UMAP_MultiPanel.png`

```python
# --- FIGURE 2 : UMAP MULTI-PANEL ---
import scanpy as sc
import matplotlib.pyplot as plt

print("-" * 70)
print("🗺️  Génération Figure 2 : UMAP Multi-Panel")
print("-" * 70)

sc.set_figure_params(dpi=150, frameon=False)
fig, axes = plt.subplots(2, 2, figsize=(14, 12))

# Panel A : Leiden
sc.pl.umap(adata_pp, color='leiden', ax=axes[0,0], show=False,
          title='A. Leiden Clusters', legend_loc='on data', size=20)

# Panel B : Diagnostic
sc.pl.umap(adata_pp, color='simple_diagnosis', ax=axes[0,1], show=False,
          title='B. Diagnosis', palette=['#2E7D32','#F57C00','#C62828'], size=20)

# Panel C : Donneurs
sc.pl.umap(adata_pp, color='donor_id', ax=axes[1,0], show=False,
          title='C. Donor', legend_loc='right margin', size=20)

# Panel D : Ancestry
if 'genetic_ancestry' in adata_pp.obs.columns:
    sc.pl.umap(adata_pp, color='genetic_ancestry', ax=axes[1,1], show=False,
              title='D. Ancestry', size=20)
else:
    sc.pl.umap(adata_pp, color='n_genes_by_counts', ax=axes[1,1], show=False,
              title='D. Quality', cmap='viridis', size=20)

plt.tight_layout()
plt.savefig(os.path.join(DIRS['FIGURES'], 'Fig2_UMAP_MultiPanel.png'), dpi=300, bbox_inches='tight')
print("✅ Sauvegardée : Fig2_UMAP_MultiPanel.png")
plt.show()
print("-" * 70)
```

# **Figure 3 : Dot Plot (Marker Genes)**

📍 **Localisation** : Après identification des marqueurs dans S2

Validation des types cellulaires via marqueurs canoniques.

**Output** : `Fig3_DotPlot_Markers.png`

```python
# --- FIGURE 3 : DOT PLOT MARKERS ---
import scanpy as sc
import matplotlib.pyplot as plt

print("-" * 70)
print("🎯 Génération Figure 3 : Dot Plot Markers")
print("-" * 70)

marker_genes = {
    'Excitatory': ['SLC17A7', 'SATB2', 'TBR1'],
    'Inhibitory': ['GAD1', 'GAD2', 'SST', 'PVALB', 'VIP'],
    'Astrocytes': ['AQP4', 'GFAP', 'SLC1A2'],
    'Oligodendrocytes': ['MBP', 'MOG', 'PLP1'],
    'OPC': ['PDGFRA', 'CSPG4', 'SOX10'],
    'Microglia': ['C1QA', 'C1QB', 'CX3CR1'],
    'Endothelial': ['CLDN5', 'FLT1', 'VWF']
}

fig, ax = plt.subplots(figsize=(10, 8))
sc.pl.dotplot(adata_pp, var_names=marker_genes, groupby='leiden',
             dendrogram=True, standard_scale='var', cmap='Reds',
             ax=ax, show=False)
plt.title('Marker Gene Expression', fontsize=14, fontweight='bold', pad=20)
plt.savefig(os.path.join(DIRS['FIGURES'], 'Fig3_DotPlot_Markers.png'), dpi=300, bbox_inches='tight')
print("✅ Sauvegardée : Fig3_DotPlot_Markers.png")
plt.show()
print("-" * 70)
```
